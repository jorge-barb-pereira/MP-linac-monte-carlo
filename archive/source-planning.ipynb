### Caracterização do Feixe de Eletrões

##### - Distriubição da Energia

Assumindo que a Energia dos eletrões que colidem com o alvo segue uma distribuição laurentziana:

> $ \displaystyle E \sim Laurentziana \space (E_0,\gamma) \begin{cases} E_0 \rightarrow \text{Valor Central} \\ \\ \gamma \rightarrow \text{Metade da Largura a Meia Altura} \end{cases}$
>
> $ \displaystyle FDP_{(E_0,\gamma)}(E) = \frac{1}{\gamma \times \pi} \times \frac{1}{\left(\frac{E-E_0}{\gamma}\right)^2+1} $
>
> $\displaystyle FDC_{(E_0,\gamma)}(E) = \frac{1}{\pi} \times \left[tan^{-1}\left(\frac{E-E_0}{\gamma}\right) + \frac{\pi}{2}\right] $

Deste modo, a distribuição segue as seguintes propriedades:

> $ \displaystyle FDP_{(E_0,\gamma)}\left(E_0-\gamma\right) = FDP_{(E_0,\gamma)}\left(E_0+\gamma \right) = \frac{FDP_{(E_0,\gamma)}(E_0)}{2} = \frac{1}{2 \times \gamma \times \pi}$

> $ \displaystyle P\left(\left[E_0- k \times \gamma \space;\space E_0 + k \times \gamma\right]\right) = FDC_{(E_0,\gamma)}\left(E_0+k \times \gamma\right) - FDC_{(E_0,\gamma)}\left(E_0-k \times \gamma\right) = \frac{1}{\pi} \times \left[tan^{-1}\left(\frac{(E_0+k \times \gamma)-E_0}{\gamma}\right) + \frac{\pi}{2}\right] - \frac{1}{\pi} \times \left[tan^{-1}\left(\frac{(E_0-k \times \gamma)-E_0}{\gamma}\right) + \frac{\pi}{2}\right]$
>
> $ \displaystyle = \frac{1}{\pi} \times \left[tan^{-1}\left(k \right) + \frac{\pi}{2} \right] - \frac{1}{\pi} \times \left[tan^{-1}\left(-k\right) + \frac{\pi}{2} \right] = \frac{1}{\pi} \times \left[tan^{-1}\left(k \right) + \frac{\pi}{2} \right] + \frac{1}{\pi} \times \left[tan^{-1}\left(k \right) - \frac{\pi}{2} \right] = \frac{2 \times tan^{-1}\left(k \right)}{\pi} $

Também é possível realizar o seguinte estudo de probabilidade em torno do valor central:

> $ \displaystyle \Delta E = \gamma \rightarrow k = \frac{1}{2} \implies P\left(\left[E_0- \frac{1}{2}\gamma \space;\space E_0 + \frac{1}{2}\gamma\right]\right) = \frac{2 \times tan^{-1}\left(\frac{1}{2} \right)}{\pi} \sim 0.295$
>
> $ \displaystyle \Delta E = 2\gamma \rightarrow k = 1 \implies P\left(\left[E_0- 1\gamma \space;\space E_0 + 1\gamma\right]\right) = \frac{2 \times tan^{-1}\left(1 \right)}{\pi} \sim 0.500$
>
> $ \displaystyle \Delta E = 4\gamma \rightarrow k = 2 \implies P\left(\left[E_0- 2\gamma \space;\space E_0 + 2\gamma\right]\right) = \frac{2 \times tan^{-1}\left(2 \right)}{\pi} \sim 0.704$
>
> $ \displaystyle \Delta E = 6\gamma \rightarrow k = 3 \implies P\left(\left[E_0- 3\gamma \space;\space E_0 + 3\gamma\right]\right) = \frac{2 \times tan^{-1}\left(3 \right)}{\pi} \sim 0.795$
>
> $ \displaystyle \Delta E = 10\gamma \rightarrow k = 5 \implies P\left(\left[E_0- 5\gamma \space;\space E_0 + 5\gamma\right]\right) = \frac{2 \times tan^{-1}\left(5 \right)}{\pi} \sim 0.874$
>
> $ \displaystyle \Delta E = 20\gamma \rightarrow k = 10 \implies P\left(\left[E_0- 10\gamma \space;\space E_0 + 10\gamma\right]\right) = \frac{2 \times tan^{-1}\left(10 \right)}{\pi} \sim 0.937$
>
> $ \displaystyle \Delta E = 30\gamma \rightarrow k = 15 \implies P\left(\left[E_0- 15\gamma \space;\space E_0 + 15\gamma\right]\right) = \frac{2 \times tan^{-1}\left(15 \right)}{\pi} \sim 0.957$

Fixando a largura a meia altura das distribuições:

> $ \displaystyle \epsilon \space (\%) = \frac{\gamma}{E_0} = 10\% \implies \gamma = \frac{E_0}{10} $
>
> $ \displaystyle E \sim Laurentziana \space (E_0,\frac{E_0}{10}) $

```python
def FDP_laurentziana (X,mu,gamma):

    FDP = []
    for x in X:
        FDP.append((1/(np.pi*gamma))*(1/(((x-mu)/gamma)**2+1)))
    return FDP

def FDC_laurenttziana (X,mu,gamma):

    FDC = []
    for x in X:
        FDC.append((1/np.pi)*(np.arctan((x-mu)/gamma)+(np.pi/2)))
    return FDC

Energias_Linac = [6, 10, 15, 18]

plt.figure(figsize=[15,7.5])
for i in range(1):

    mu = Energias_Linac[i]
    gamma = mu/10
    k=10
    Energias = np.arange(mu-k*gamma, mu+k*gamma,0.001)
    FDP_Energy = FDP_laurentziana(Energias, mu, gamma)

    plt.plot(Energias,FDP_Energy),plt.xlabel('Energia do Eletrão (MeV)'),plt.ylabel('Função Densidade de Probabilidade (MeV ⁻¹)')

bins = 35
for i in range(1):

    mu = Energias_Linac[i]
    gamma = mu/10
    k=10
    dE = 2*k*gamma/(bins-1)
    Energias = np.arange(mu-k*gamma, mu+k*gamma + dE, dE)
    FDP_Energy = FDP_laurentziana(Energias, mu, gamma)

    plt.bar(Energias,FDP_Energy,width=dE*0.9,alpha = 0.5),plt.xlabel('Energia do Eletrão (MeV)'),plt.ylabel('Função Densidade de Probabilidade (MeV ⁻¹)')

plt.legend(['      Energia Central = 6 Mev    \nLargura a Meia Altura = 1,2 MeV'])#,'Energia Média = 10 Mev','Energia Média = 15 Mev','Energia Média = 18 Mev'])
plt.title('Distribuição Energética dos Eletrões')
plt.show()
```

##### - Distribuição Espacial

Através de um Sistema de Coordenadas Polares é possível definir uma posição no plano XY da seguinte forma:

> $ \displaystyle  P \equiv(x,y) \Leftrightarrow P \equiv (r,\theta) : \begin{cases} r \equiv \sqrt{x^2 + y^2} \\ \\ \theta \equiv tan^{-1} \left(\frac{y}{x}\right)\end{cases}$

Assumindo que a Coordenada Radial dos eletrões que colidem com o alvo segue uma distribuição laurentziana:

> $ \displaystyle R \sim Laurentziana \space \rm I\!R^{+}_0 \space (R_0,\gamma) \begin{cases} R_0 \rightarrow \text{Valor Central} \\ \\ \gamma \rightarrow \text{Metade da Largura a Meia Altura} \end{cases}$
>
> $ \displaystyle FDP(R) = \frac{1}{\gamma \times \pi} \times \frac{1}{(\frac{R-R_0}{\gamma})^2+1} $
>
> $\displaystyle FDC(R) = \frac{1}{\pi} \times \left[tan^{-1}(\frac{R-R_0}{\gamma}) + \frac{\pi}{2}\right] $

Contudo, é necessário ter em atenção que a Coordenada Radial dos eletrões, não pode ser negativa (por definição) e como tal o Domínio da sua Função Densidade de Probabilidade resume-se a ${\rm I\!R_{0}^{+}}$:

> $\displaystyle R \geq 0 \implies FDP^{'}_{(R_0,\gamma)} \space (R):{\rm I\!R_{0}^{+}} \rightarrow [0:1] \implies FDC^{'}_{(R_0,\gamma)} \space (R) = \int_{0}^{R} FDP^{'}_{(R_0,\gamma)} \space (R^{'}) \space d(R^{'}) \space : \space FDC^{'}_{(R_0,\gamma)} \space (+\infty) = 1$
>
> $\displaystyle FDP^{'}_{(R_0,\gamma)} \space (R) = K \times \frac{1}{(\frac{R-R_0}{\gamma})^2+1} : K > \frac{1}{\gamma \times \pi} $

É possível determinar facilmente o valor da nova constante de normalização $K$ através de manipulações algébricas:

> $\displaystyle FDC_{(R_0,\gamma)} \space (+\infty) = 1 \Leftrightarrow \int_{-\infty}^{+\infty} FDP_{(R_0,\gamma)} \space (R^{'}) \space dR^{'} = 1 \Leftrightarrow \int_{-\infty}^{0} FDP_{(R_0,\gamma)} \space (R^{'}) \space dR^{'} + \int_{0}^{+\infty} FDP_{(R_0,\gamma)} \space (R^{'}) \space dR^{'} = 1 $
>
> $\displaystyle \Leftrightarrow FDC_{(R_0,\gamma)} \space (0) + \int_{0}^{+\infty} FDP_{(R_0,\gamma)} \space (R^{'}) \space dR^{'} = 1 \Leftrightarrow  \int_{0}^{+\infty} FDP_{(R_0,\gamma)} \space (R^{'}) \space dR^{'} = 1 - FDC_{(R_0,\gamma)} \space (0) $
>
> $\displaystyle \Leftrightarrow \frac{1}{\gamma \times \pi} \int_{0}^{+\infty}  \frac{1}{(\frac{R-R_0}{\gamma})^2+1} \space d(R^{'}) = 1 - FDC_{(R_0,\gamma)} \space (0) \Leftrightarrow \int_{0}^{+\infty} \frac{1}{(\frac{R-R_0}{\gamma})^2+1} \space dR^{'} = \gamma \times \pi \times [ 1 - FDC_{(R_0,\gamma)} \space (0) ]$

> $\displaystyle FDC^{'}_{(R_0,\gamma)} \space (+\infty) = 1 \Leftrightarrow \int_{0}^{+\infty} FDP^{'}_{(R_0,\gamma)} \space (R^{'}) \space dR^{'} = 1 \Leftrightarrow K \int_{0}^{+\infty} \frac{1}{(\frac{R-R_0}{\gamma})^2+1} dR^{'} = 1 \Leftrightarrow K \times \gamma \times \pi \times [ 1 - FDC_{(R_0,\gamma)} \space (0) ] = 1 $
>
> $\displaystyle \Leftrightarrow K = \frac{1}{\gamma \times \pi} \times \frac{1}{1 - FDC_{(R_0,\gamma)} \space (0)}$

Assim chegamos à Função Densidade de Probabilidades e à Função Distribuição Cumulativa quem descrevem satisfatoriamente a Coordenada Radial dos eletrões:

> $\displaystyle FDP^{'}_{(R_0,\gamma)} \space (R) = \frac{1}{\gamma \times \pi} \times \frac{1}{1 - FDC_{(R_0,\gamma)} \space (0)} \times \frac{1}{(\frac{r-R_0}{\gamma})^2+1} = \frac{1}{1 - FDC_{(R_0,\gamma)} \space (0)} \times FDP_{(R_0,\gamma)} \space (R)$

> $\displaystyle FDC^{'}_{(R_0,\gamma)} \space (R) = \int_{0}^{R} FDP^{'}_{(R_0,\gamma)} \space (R^{'}) \space dR^{'} = \frac{1}{\gamma \times \pi} \times \frac{1}{1 - FDC_{(R_0,\gamma)} \space (0)} \int_{0}^{R} \frac{1}{(\frac{R^{'}-R_0}{\gamma})^2+1}  \space dR^{'} $
>
> $\displaystyle= \frac{1}{1 - FDC_{(R_0,\gamma)} \space (0)} \times \left[FDC_{(R_0,\gamma)} \space (R) - FDC_{(R_0,\gamma)} \space (0)\right]$

Fixando a média e a largura a meia altura em valores típicos:

> $\displaystyle \begin{cases} R_0 = 0 \space cm \\ \\ \gamma = 0.10 \space cm \end{cases} \implies FDP'(R) = \frac{1}{1-FDC_{(R_0,\gamma)}(0)} \times FDP_{(R_0,\gamma)}(R) = \frac{1}{0.50} \times FDP_{(R_0,\gamma)}(R) = \frac{2}{\gamma \times \pi} \times \frac{1}{(\frac{R-R_0}{\gamma})^2+1}$

Deste modo, a distribuição segue as seguintes propriedades:

> $ \displaystyle FDP^{'}_{(R_0,\gamma)}\left(R_0+\gamma\right) = \frac{FDP^{'}_{(R_0,\gamma)}(R_0)}{2} = \frac{1}{\gamma \times \pi}$

> $ \displaystyle P\left(\left[0 \space;\space R_0 + k \times \gamma\right]\right) = FDC_{(R_0,\gamma)}^{'}\left(R_0+k \times \gamma\right) = \frac{1}{1 - FDC_{(R_0,\gamma)} \space (0)} \times \left[FDC_{(R_0,\gamma)} \space (R_0 + k \times \gamma) - FDC_{(R_0,\gamma)} \space (0)\right]$
>
> $ \displaystyle = \frac{1}{0.5} \times \left[FDC_{(R_0,\gamma)} \space (R_0 + k \times \gamma) -0.5\right] = 2 \times \left[ \frac{1}{\pi} \times \left[tan^{-1}\left(\frac{(R_0+k \times \gamma)-R_0}{\gamma}\right) + \frac{\pi}{2}\right] -0.5\right] = \frac{2}{\pi} \times \left[tan^{-1}\left(k \right) + \frac{\pi}{2} \right] - 1  = \frac{2 \times tan^{-1}\left(k \right)}{\pi} $

Assumindo que a Posição Angular dos eletrões que colidem com o alvo segue uma distribuição uniforme:

> $ \displaystyle \Theta \sim Uniforme $
>
> $ \displaystyle FDP(\Theta) = \begin{cases} 0 \text{ ,  se θ < 0}\\ \\ \frac{1}{2 \pi} \text{,  se 0 ≤ θ < }2\pi \\ \\ 0 \text{, se 0 ≥ }2\pi \end{cases}$
>
> $\displaystyle FDC(\Theta) = \begin{cases} 0 \text{ ,  se θ < 0}\\ \\ \frac{\theta}{2 \pi} \text{,  se 0 ≤ θ < }2\pi \\ \\ 1 \text{, se 0 ≥ }2\pi \end{cases} $

```python
def FDP_laurentziana_trunc (X,mu,gamma):

    FDP = []
    for x in X:
        FDP.append((2/(np.pi*gamma))*(1/(((x-mu)/gamma)**2+1)))
    return FDP

mu = 0 #cm
gamma = 0.1 #cm
k=10

plt.figure(figsize=[20,7.5])
plt.subplot(1,2,1)
R = np.arange(0, mu+k*gamma,0.001)
FDP_R = FDP_laurentziana_trunc(R, mu, gamma)
plt.plot(R,FDP_R),plt.xlabel('Posição R do Eletrão (cm)'),plt.ylabel('Função Densidade de Probabilidade (cm ⁻¹)')

bins = 35
dE = 1*k*gamma/(bins-1)
X = np.arange(0, mu+k*gamma + dE, dE)
FDP_X = FDP_laurentziana_trunc(X, mu, gamma)
plt.bar(X,FDP_X,width=dE*0.9,alpha = 0.5),plt.xlabel('Posição R do Eletrão (cm)'),plt.ylabel('Função Densidade de Probabilidade (cm ⁻¹)')

def FDP_uniforme (X_FDP,a,b):

    k = 1/(b-a)

    FDP = []
    for x in X_FDP:
        FDP.append(k)

    return FDP

plt.subplot(1,2,2)
Theta = np.arange(0, 2*np.pi,0.001)
FDP_Theta = FDP_uniforme(Theta,0,2*np.pi)
plt.plot(Theta,FDP_Theta),plt.xlabel('Posição θ do Eletrão (rad)'),plt.ylabel('Função Densidade de Probabilidade (rad ⁻¹)')

bins = 25
dTheta = 2*np.pi/(bins-1)
Theta = np.arange(0, 2*np.pi + dTheta, dTheta)
FDP_Theta = FDP_uniforme(Theta,0,2*np.pi)
plt.bar(Theta,FDP_Theta,width=dTheta*0.9,alpha = 0.5),plt.xlabel('Posição θ do Eletrão (rad)'),plt.ylabel('Função Densidade de Probabilidade (rad ⁻¹)')
plt.show()

```
